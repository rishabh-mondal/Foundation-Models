```python
import os
import numpy as np
import pandas as pd
import geopandas as gpd
from shapely.geometry import Polygon

patch_size_pix = 128
patch_size_m = patch_size_pix * 10

def get_patch_degree_size(lat_center, patch_size_m=1280):
    deg_per_meter_lat = 1 / 111320
    deg_per_meter_lon = 1 / (111320 * np.cos(np.deg2rad(lat_center)))
    patch_height_deg = patch_size_m * deg_per_meter_lat
    patch_width_deg  = patch_size_m * deg_per_meter_lon
    return patch_height_deg, patch_width_deg

def get_patch_bounds(lat_center, lon_center, patch_height_deg, patch_width_deg):
    min_lon = lon_center - patch_width_deg / 2
    max_lon = lon_center + patch_width_deg / 2
    min_lat = lat_center - patch_height_deg / 2
    max_lat = lat_center + patch_height_deg / 2
    return min_lon, min_lat, max_lon, max_lat

def pixel_to_latlon(x, y, min_lon, min_lat, max_lon, max_lat, img_w=128, img_h=128):
    # Map (0,0) to (min_lon, max_lat), (img_w-1, img_h-1) to (max_lon, min_lat)
    lon = min_lon + (x / (img_w - 1)) * (max_lon - min_lon)
    lat = max_lat - (y / (img_h - 1)) * (max_lat - min_lat)
    return lon, lat


records = []
for idx, row in df.iterrows():
    lat_center = row['actual_lat_center']
    lon_center = row['actual_lon_center']
    patch_id = row['patch_id']
    label_path = os.path.join(labels_dir, f"{patch_id}.txt")
    if not os.path.exists(label_path) or pd.isna(lat_center) or pd.isna(lon_center):
        continue
    patch_height_deg, patch_width_deg = get_patch_degree_size(lat_center)
    min_lon, min_lat, max_lon, max_lat = get_patch_bounds(lat_center, lon_center, patch_height_deg, patch_width_deg)
    with open(label_path) as f:
        for line in f:
            vals = list(map(float, line.strip().split()))
            cls = int(vals[0])
            points = np.array(vals[1:]).reshape(4, 2)
            # Use (img_w-1), (img_h-1) for correct pixel index mapping
            points_pix = points * (patch_size_pix - 1)
            coords = [pixel_to_latlon(x, y, min_lon, min_lat, max_lon, max_lat, patch_size_pix, patch_size_pix) for x, y in points_pix]
            geom = Polygon(coords)
            records.append({
                'patch_id': patch_id,
                'class': cls,
                'geometry': geom
            })

labels_gdf = gpd.GeoDataFrame(records, crs="EPSG:4326")
labels_gdf.to_file(f"{state}_labels_polygons.geojson", driver="GeoJSON")

print(labels_gdf.head())
print(f"Total label polygons: {len(labels_gdf)}")

#count the  unique patch id
unique_patch_ids = df['patch_id'].nunique()
print(f"Total unique patch IDs: {unique_patch_ids}")
```

```python
import numpy as np
import pandas as pd
import geopandas as gpd
from shapely.geometry import Polygon

# ---- CONFIG ----
STATE = "uttar_pradesh"  # change for other states
# ----------------

def get_patch_degree_size(lat_center, patch_size_m=1280):
    deg_per_meter_lat = 1 / 111320
    deg_per_meter_lon = 1 / (111320 * np.cos(np.deg2rad(lat_center)))
    patch_height_deg = patch_size_m * deg_per_meter_lat
    patch_width_deg  = patch_size_m * deg_per_meter_lon
    return patch_height_deg, patch_width_deg

def get_patch_bounds(lat_center, lon_center, patch_height_deg, patch_width_deg):
    min_lon = lon_center - patch_width_deg / 2
    max_lon = lon_center + patch_width_deg / 2
    min_lat = lat_center - patch_height_deg / 2
    max_lat = lat_center + patch_height_deg / 2
    return min_lon, min_lat, max_lon, max_lat

assert labels_gdf.crs and labels_gdf.crs.to_epsg() == 4326, "labels_gdf must be EPSG:4326"

# 1) Build patch footprints
patch_records = []
for _, r in df.dropna(subset=["actual_lat_center","actual_lon_center","patch_id"]).iterrows():
    latc = float(r["actual_lat_center"]); lonc = float(r["actual_lon_center"]); pid = r["patch_id"]
    hdeg, wdeg = get_patch_degree_size(latc)
    minx, miny, maxx, maxy = get_patch_bounds(latc, lonc, hdeg, wdeg)
    poly = Polygon([(minx, miny), (minx, maxy), (maxx, maxy), (maxx, miny)])
    patch_records.append({"patch_id": pid, "geometry": poly})
patch_gdf = gpd.GeoDataFrame(patch_records, crs="EPSG:4326")

# 2) Labels unique id
labels = labels_gdf.copy()
labels["label_id"] = np.arange(len(labels), dtype=int)

# 3) Spatial join: polygons ↔ patches
joined = gpd.sjoin(
    labels[["label_id","patch_id","class","geometry"]],
    patch_gdf[["patch_id","geometry"]].rename(columns={"patch_id":"patch_id_foot"}),
    how="left",
    predicate="intersects"
)

# 4) Count distinct patches per label
patch_counts = (joined.groupby("label_id")["patch_id_foot"]
                      .nunique(dropna=True)
                      .rename("n_patches")
                      .reset_index())

labels_with_counts = labels.merge(patch_counts, on="label_id", how="left")
labels_with_counts["n_patches"] = labels_with_counts["n_patches"].fillna(0).astype(int)

# 5) Two buckets only
labels_with_counts["common_bin"] = np.where(labels_with_counts["n_patches"] <= 1,
                                            "no_common", "common")

# 6) Summary: label counts
label_summary = labels_with_counts["common_bin"].value_counts().reindex(["no_common","common"]).fillna(0).astype(int)
print("Counts by bin (labels):")
print(label_summary.to_string())

# 7) Summary: unique patch_id counts
patch_summary = (labels_with_counts.groupby("common_bin")["patch_id"]
                 .nunique()
                 .reindex(["no_common","common"])
                 .fillna(0).astype(int))
print("\nCounts by bin (unique patch_ids):")
print(patch_summary.to_string())

# 8) Save master CSV
out_csv = f"{STATE}_labels_commonness_summary_two_bins.csv"
labels_with_counts[["label_id","patch_id","class","n_patches","common_bin"]].to_csv(out_csv, index=False)
print(f"\nSaved master CSV: {out_csv}")
```

```python
from pathlib import Path

base = Path("/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/iclr_2026_processed_data/final_data")
splits = ["train", "val", "test"]

def count_instances_in_file(p: Path) -> int:
    try:
        with open(p, "r", encoding="utf-8", errors="ignore") as f:
            return sum(1 for ln in f if ln.strip())
    except FileNotFoundError:
        return 0

def count_instances(lbl_files):
    return sum(count_instances_in_file(p) for p in lbl_files)

def stemset(paths):
    return {p.stem for p in paths}

def top_k_instance_labels(lbl_files, k=5):
    counts = [(p, count_instances_in_file(p)) for p in lbl_files]
    counts.sort(key=lambda x: (-x[1], x[0].name))   # by count desc, then filename
    return counts[:k]

grand_imgs = grand_lbls = grand_insts = 0
grand_pos = grand_neg = 0

# to compute region-wide and global top-5s
global_label_files = []

print("===== Dataset Summary by Region & Split =====")
for region_dir in sorted([d for d in base.iterdir() if d.is_dir()]):
    region = region_dir.name
    reg_imgs = reg_lbls = reg_insts = 0
    reg_pos = reg_neg = 0
    region_lbl_files_all = []

    print(f"\n[Region] {region}")
    for split in splits:
        img_dir = region_dir / split / "images"
        lbl_dir = region_dir / split / "labels"

        img_files = list(img_dir.glob("*.png")) if img_dir.exists() else []
        lbl_files = list(lbl_dir.glob("*.txt")) if lbl_dir.exists() else []

        region_lbl_files_all.extend(lbl_files)
        global_label_files.extend(lbl_files)

        inst_cnt = count_instances(lbl_files)

        img_stems = stemset(img_files)
        lbl_stems = stemset(lbl_files)

        pos_imgs = len(img_stems & lbl_stems)  # positive = with labels
        neg_imgs = len(img_stems - lbl_stems)  # negative = without labels

        print(f"  {split.capitalize():5} -> Images: {len(img_files):5} | Labels: {len(lbl_files):5} | "
              f"Instances: {inst_cnt:6} | Pos: {pos_imgs:5} | Neg: {neg_imgs:5}")

        # ---- Top-5 label files by instances (this split) ----
        if lbl_files:
            top5 = top_k_instance_labels(lbl_files, k=5)
            print("      Top-5 label files (by instances):")
            for p, c in top5:
                print(f"        {p.name:30}  instances={c}")
        else:
            print("      Top-5 label files (by instances):  - none -")

        reg_imgs += len(img_files)
        reg_lbls += len(lbl_files)
        reg_insts += inst_cnt
        reg_pos  += pos_imgs
        reg_neg  += neg_imgs

    print(f"  -- Region totals -> Images: {reg_imgs} | Labels: {reg_lbls} | Instances: {reg_insts} "
          f"| Pos: {reg_pos} | Neg: {reg_neg}")

    # ---- Top-5 for the whole region (across train/val/test) ----
    if region_lbl_files_all:
        region_top5 = top_k_instance_labels(region_lbl_files_all, k=5)
        print("  -- Region-wide Top-5 label files (by instances):")
        for p, c in region_top5:
            # show split folder for clarity
            try:
                split_name = p.parents[1].name  # .../<split>/labels/<file>.txt
            except Exception:
                split_name = "?"
            print(f"     [{split_name:4}] {p.name:30}  instances={c}")

    grand_imgs += reg_imgs
    grand_lbls += reg_lbls
    grand_insts += reg_insts
    grand_pos  += reg_pos
    grand_neg  += reg_neg

print("\n===== GRAND TOTALS =====")
print(f"Images: {grand_imgs} | Labels: {grand_lbls} | Instances: {grand_insts} | Pos: {grand_pos} | Neg: {grand_neg}")

# ---- Global Top-5 (across all regions & splits) ----
if global_label_files:
    global_top5 = top_k_instance_labels(global_label_files, k=5)
    print("\n===== GLOBAL Top-5 label files (by instances) =====")
    for p, c in global_top5:
        # include region/split context
        try:
            split_name = p.parents[1].name  # .../<split>/labels
            region_name = p.parents[3].name # .../<region>/<split>/labels
        except Exception:
            split_name, region_name = "?", "?"
        print(f"  [{region_name} / {split_name:4}] {p.name:30}  instances={c}")
else:
    print("\n===== GLOBAL Top-5 label files (by instances) =====\n  - none -")
```
